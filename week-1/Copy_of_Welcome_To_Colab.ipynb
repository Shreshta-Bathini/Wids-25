# Week One : Implement tokenization & stemming using Python + NLTK




```
import nltk
nltk.download('punkt')
nltk.download('stopwords')

import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')


from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

text = "Natural Language Processing is very interesting!"

# Tokenization
tokens = word_tokenize(text)

# Remove stopwords
stop_words = set(stopwords.words('english'))
tokens = [w for w in tokens if w.lower() not in stop_words and w.isalpha()]

# Stemming
stemmer = PorterStemmer()
stems = [stemmer.stem(w) for w in tokens]

print("Tokens:", tokens)
print("Stems:", stems)
```

# Using spaCy

```
import spacy
nlp = spacy.load("en_core_web_sm")

doc = nlp("Natural Language Processing is very interesting!")

tokens = [token.text for token in doc if not token.is_stop and token.is_alpha]
lemmas = [token.lemma_ for token in doc if not token.is_stop and token.is_alpha]

print("Tokens:", tokens)
print("Lemmas:", lemmas)
```

# TF-IDF Sentiment Classifier (IMDb / SST-2)

```
from datasets import load_dataset

dataset = load_dataset("imdb")
train_texts = dataset["train"]["text"]
train_labels = dataset["train"]["label"]
```

```
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1,2),
    stop_words='english'
)

X_train = vectorizer.fit_transform(train_texts)
```

```
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(X_train, train_labels)
```

```
from sklearn.metrics import accuracy_score

test_texts = dataset["test"]["text"]
test_labels = dataset["test"]["label"]

X_test = vectorizer.transform(test_texts)
preds = model.predict(X_test)

print("Accuracy:", accuracy_score(test_labels, preds))
```
